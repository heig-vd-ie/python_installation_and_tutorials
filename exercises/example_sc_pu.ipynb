```python
import numpy as np
```

```python
# Valeur nominal
P_n = 10
U_n = np.array([10, 5, 20, 10, 5])

# Résistance de ligne
R_l = np.array([1, 2, 70, 10, 7])
# Tension d'entree (HT)
U_mes_1 = [9.8]
U_mes_2 = []
```

```python
# Calcul d'autre valeur nominales
R_n = U_n**2/P_n
I_n = P_n/U_n
# Résistance en pu ramener à la HT
R_pu = R_l/R_n*R_n[0]
# Résistance totale
R_tot = sum(R_pu)
```

```python
# Courant de court-circuit en pu
i_cc_pu = U_mes_1[0]/R_tot
# Courant de court-circuit en pu
i_cc = i_cc_pu*I_n
# Courant de court-circuit en A
u_l = i_cc*R_l
print("Courant de court-circuit [A]: ", list(np.around(i_cc,2)))
print("Chute de tension dans les ligne [V]: ", list(np.around(u_l,2)))
```

```python
# Calcul des tension mesurées en V au noeud avant et apres les lignes de transmission
for i in range(len(U_n)):
    U_mes_2.append(U_mes_1[i] - u_l[i])
    if i != len(U_n)-1:
        U_mes_1.append(U_mes_2[i]*U_n[i+1]/U_n[i])
print("Tension mesurée [V]")
print("Debut de ligne: " , list(np.around(np.array(U_mes_1),2)))
print("Fin de ligne: ",list(np.around(np.array(U_mes_2),2)))
```
